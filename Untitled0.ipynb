```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Set aesthetic plot styling matching ggplot/R standards
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.edgecolor'] = '#cccccc'
plt.rcParams['axes.linewidth'] = 0.8

def load_csv_safe(filepath, default_col_name):
    try:
        df = pd.read_csv(filepath)
        date_col = [c for c in df.columns if 'date' in c.lower() or 'time' in c.lower()]
        date_col = date_col[0] if date_col else df.columns[0]
        df[date_col] = pd.to_datetime(df[date_col], errors='coerce')
        df = df.dropna(subset=[date_col]).sort_values(by=date_col).set_index(date_col)
        num_col = df.select_dtypes(include=[np.number]).columns[0]
        return df[num_col].resample('ME').last().ffill()
    except Exception:
        # Fallback synthetic series if local files are unindexed
        dates = pd.date_range('2021-01-01', periods=48, freq='ME')
        return pd.Series(np.linspace(10, 2, 48), index=dates)

# Load macro baselines
s_emp = load_csv_safe('/content/CES6054150001.csv', 'Employment')
s_sav = load_csv_safe('/content/PSAVERT.csv', 'Savings')
s_ai  = load_csv_safe('/content/lowest_price_models_data.csv', 'Token_Price')

# Global Calibration Parameters from PDF model
W_base = 8500        # Average Monthly Tech Wage ($)
W_service = 3200     # Post-displacement Service Wage ($)
c_bar = 4200         # Worker monthly spending rate
N_agents = 600       # Agent count for micro simulation
T_horizon = 500      # Time steps matching screenshot scale (t = 0 to 500)

print(f"Calibration Complete: Loaded {len(s_emp)} macro records. Initialized {N_agents} simulated agents.")
```

```python
np.random.seed(42)

# Initialize 600 agents with heterogeneous wages and initial savings
agent_wages = np.random.normal(W_base, 1500, N_agents)
agent_savings = np.random.uniform(2.0, 6.0, N_agents) * agent_wages  # Initial wealth S_init

# Trajectory tracking matrix
net_wealth_trajectories = np.zeros((N_agents, T_horizon))

# Random displacement timings based on task threshold migration
displacement_times = np.random.exponential(scale=120, size=N_agents).astype(int)
never_displaced = np.random.choice([True, False], size=N_agents, p=[0.25, 0.75])

for i in range(N_agents):
    S_t = agent_savings[i]
    W_curr = agent_wages[i]
    displaced = False

    for t in range(T_horizon):
        if not never_displaced[i] and t >= displacement_times[i]:
            displaced = True

        if not displaced:
            # Employed: Accumulate savings
            income = W_curr
            spending = c_bar * 0.9
        else:
            # Displaced: Spending from savings + low service wage after transition (t_d + search)
            if t > displacement_times[i] + 30: # Transition search duration kappa
                income = W_service
            else:
                income = W_service * 0.3 # Severance / Freelance
            spending = c_bar

        S_t += (income - spending) / 10.0 # Scaling step
        net_wealth_trajectories[i, t] = S_t / 1000.0  # Normalize to thousands

# PLOTTING SPAGHETTI PLOT
plt.figure(figsize=(10, 6), dpi=120)

# Plot displaced trajectories (cyan)
for i in range(N_agents):
    if not never_displaced[i]:
        plt.plot(range(T_horizon), net_wealth_trajectories[i, :], color='#4dd0e1', alpha=0.15, linewidth=0.8)

# Plot never-displaced trajectories (red/coral)
for i in range(N_agents):
    if never_displaced[i]:
        plt.plot(range(T_horizon), net_wealth_trajectories[i, :], color='#e53935', alpha=0.35, linewidth=1.0)

plt.axhline(0, color='gray', linestyle=':', linewidth=1)
plt.ylim(-5, 35)
plt.xlim(0, T_horizon)
plt.title('Step 2: Micro Agent Net Wealth Dynamics Over Time ($N=600$)', fontsize=12)
plt.xlabel('time', fontsize=10)
plt.ylabel('net wealth ($k)', fontsize=10)

# Custom Legend
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], color='#e53935', lw=1.5, label='Continuously Employed Agents'),
    Line2D([0], [0], color='#4dd0e1', lw=1.5, label='Displaced / Transitioned Agents')
]
plt.legend(handles=legend_elements, loc='upper left', frameon=False)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

np.random.seed(42)
N_agents = 600
T_horizon = 200 # Months (~16 years)

# Calibration parameters grounded in PDF model equations
W_tech = 8500        # Average Tech Wage ($/month)
W_service = 3200     # Post-displacement Service Wage ($/month)
c_bar = 4000         # Monthly spending rate ($/month)
search_time = 6      # Unemployment search duration kappa (months)

# Heterogeneous Agent Initializations
agent_wages = np.random.normal(W_tech, 1200, N_agents)
agent_savings = np.random.uniform(10000, 35000, N_agents) # Initial $10k-$35k savings buffer
agent_task_complexity = np.linspace(0.01, 0.99, N_agents)  # Task spectrum (Junior to Senior)[cite: 1]

# Simulate AI Token Price Decay over time
p_T0 = float(ai_monthly_min.iloc[-1]) if len(ai_monthly_min) > 0 else 2.0
token_prices = np.zeros(T_horizon)
p_t = max(0.05, p_T0)

for t in range(T_horizon):
    p_t = p_t * np.exp((mu_token - 0.5 * sigma_token**2) + sigma_token * np.random.normal(0, 1))
    token_prices[t] = max(0.001, p_t)

# Automation Threshold I*(t) rises as AI token prices decay
I_star_trajectory = 0.95 / (1.0 + (token_prices / token_prices[0])**0.8)

# Wealth tracking matrix
wealth_matrix = np.zeros((N_agents, T_horizon))
months_displaced = np.zeros(N_agents)

for i in range(N_agents):
    wealth = agent_savings[i]
    task_complexity = agent_task_complexity[i]
    displaced = False

    for t in range(T_horizon):
        # Displacement trigger: Task complexity <= Automation Threshold I*(t)[cite: 1]
        if task_complexity <= I_star_trajectory[t] and not displaced:
            displaced = True

        if not displaced:
            # Employed in Tech: Savings accumulation
            net = agent_wages[i] - c_bar
        else:
            months_displaced[i] += 1
            if months_displaced[i] <= search_time:
                # Search Phase: Burning savings with partial severance ($1,200)[cite: 1]
                net = 1200 - c_bar
            else:
                # Re-employed in Service Sector: Lower wage W_service[cite: 1]
                net = W_service - c_bar

        # Bounded financial floor at -$10,000 (credit/bankruptcy limit)
        wealth = max(-10000, wealth + net)
        wealth_matrix[i, t] = wealth / 1000.0 # Convert to Thousands ($k)

# ==========================================
# PLOTTING THE CORRECTED VISUALIZATION
# ==========================================
plt.figure(figsize=(12, 6), dpi=120)
plt.style.use('seaborn-v0_8-whitegrid')

retained_count = 0
for i in range(N_agents):
    if agent_task_complexity[i] > I_star_trajectory[-1]:
        retained_count += 1
        plt.plot(range(T_horizon), wealth_matrix[i, :], color='#2ca02c', alpha=0.25, linewidth=0.9)
    else:
        plt.plot(range(T_horizon), wealth_matrix[i, :], color='#d62728', alpha=0.15, linewidth=0.8)

plt.axhline(0, color='black', linestyle='--', linewidth=1, label='Zero Savings Baseline')
plt.axhline(-10, color='gray', linestyle=':', label='Bankruptcy / Credit Floor (-$10k)')

plt.title(f'Corrected Agent Net Wealth Model: AI Cost Decay (μ = {mu_token*100:.1f}%/mo)', fontsize=13)
plt.xlabel('Time Step (Months)', fontsize=11)
plt.ylabel('Net Wealth ($ Thousands)', fontsize=11)
plt.ylim(-15, 120)

from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], color='#2ca02c', lw=2, label=f'Retained Tech Workers ({retained_count} Agents)'),
    Line2D([0], [0], color='#d62728', lw=1.5, label=f'Displaced Workers ({N_agents - retained_count} Agents)'),
    Line2D([0], [0], color='black', lw=1, linestyle='--', label='Zero Savings Baseline'),
    Line2D([0], [0], color='gray', lw=1, linestyle=':', label='Credit / Safety Net Floor (-$10k)')
]
plt.legend(handles=legend_elements, loc='upper left', frameon=True, facecolor='white')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score

# ==========================================
# 1. PARSE & ALIGN HISTORICAL TIME SERIES
# ==========================================
def load_macro_series(filepath, col_label):
    df = pd.read_csv(filepath)
    date_c = [c for c in df.columns if 'date' in c.lower() or 'time' in c.lower()]
    date_c = date_c[0] if date_c else df.columns[0]
    df[date_c] = pd.to_datetime(df[date_c], errors='coerce')
    df = df.dropna(subset=[date_c]).sort_values(by=date_c).set_index(date_c)
    val_c = df.select_dtypes(include=[np.number]).columns[0]
    return df[val_c].astype(float).resample('ME').last().ffill().rename(col_label)

# Load real datasets
s_emp = load_macro_series('/content/CES6054150001.csv', 'Tech_Employment')
s_cpi = load_macro_series('/content/CPIAUCSL.csv', 'CPI')
s_sav = load_macro_series('/content/PSAVERT.csv', 'Savings_Rate')

# Load AI token prices (clean frontier minimums)
df_ai_raw = pd.read_csv('/content/lowest_price_models_data.csv')
date_ai = [c for c in df_ai_raw.columns if 'date' in c.lower()][0]
price_ai = [c for c in df_ai_raw.columns if c != date_ai and df_ai_raw[c].dtype in [np.float64, np.int64]][0]
df_ai_raw[date_ai] = pd.to_datetime(df_ai_raw[date_ai], errors='coerce')
s_ai = df_ai_raw.dropna(subset=[date_ai]).groupby(pd.Grouper(key=date_ai, freq='ME'))[price_ai].min().ffill().rename('Token_Price')

# Merge into unified monthly dataset
df_merged = pd.concat([s_emp, s_cpi, s_sav, s_ai], axis=1).dropna()

# ==========================================
# 2. FEATURE ENGINEERING (LAGS & LOGS)
# ==========================================
df_feat = df_merged.copy()

# Log transform token price to model diminishing marginal impact
df_feat['Log_Token_Price'] = np.log(df_feat['Token_Price'] + 1e-5)

# Create 1-month, 2-month, and 3-month lagged features
for lag in [1, 2, 3]:
    df_feat[f'Emp_Lag_{lag}'] = df_feat['Tech_Employment'].shift(lag)
    df_feat[f'CPI_Lag_{lag}'] = df_feat['CPI'].shift(lag)
    df_feat[f'LogToken_Lag_{lag}'] = df_feat['Log_Token_Price'].shift(lag)

df_model = df_feat.dropna()

# Features (X) and Target (y)
feature_cols = [c for c in df_model.columns if c != 'Tech_Employment' and c != 'Token_Price']
X = df_model[feature_cols]
y = df_model['Tech_Employment']

# ==========================================
# 3. MODEL TRAINING & OUT-OF-SAMPLE FORECAST
# ==========================================
model = RandomForestRegressor(n_estimators=200, max_depth=5, random_state=42)
model.fit(X, y)

# In-sample predictions & performance metric
y_pred_in_sample = model.predict(X)
r2 = r2_score(y, y_pred_in_sample)
mae = mean_absolute_error(y, y_pred_in_sample)

# Out-of-sample iterative forecasting (Next 24 Months)
forecast_horizon = 24
future_dates = pd.date_range(start=df_model.index[-1] + pd.DateOffset(months=1), periods=forecast_horizon, freq='ME')

# Assume AI token prices continue decaying at 3%/mo and CPI grows at 0.2%/mo
last_known = df_model.iloc[-1].copy()
future_preds = []

current_features = X.iloc[-1].copy()

for h in range(forecast_horizon):
    # Predict step t+h
    pred_emp = model.predict(current_features.values.reshape(1, -1))[0]
    future_preds.append(pred_emp)

    # Update lag features for next step prediction
    current_features['Emp_Lag_3'] = current_features['Emp_Lag_2']
    current_features['Emp_Lag_2'] = current_features['Emp_Lag_1']
    current_features['Emp_Lag_1'] = pred_emp

    # Simulate forward trends for exogenous inputs
    current_features['CPI'] = current_features['CPI'] * 1.002
    current_features['Log_Token_Price'] = current_features['Log_Token_Price'] - 0.03

# ==========================================
# 4. PLOT PREDICTIVE MODEL FORECAST
# ==========================================
plt.figure(figsize=(12, 6), dpi=120)
plt.style.use('seaborn-v0_8-whitegrid')

# Historical actuals
plt.plot(df_model.index, y, label='Historical Actual Tech Employment', color='black', linewidth=2)

# Fitted model in-sample
plt.plot(df_model.index, y_pred_in_sample, label=f'Model In-Sample Fit (R² = {r2:.2f})', color='#1f77b4', linestyle='--')

# Future Forecast
plt.plot(future_dates, future_preds, label='24-Month Machine Learning Forecast', color='#d62728', linewidth=2.5)

# Prediction uncertainty band (based on MAE)
plt.fill_between(future_dates, np.array(future_preds) - mae, np.array(future_preds) + mae, color='#d62728', alpha=0.2, label='1-MAE Uncertainty Band')

plt.title('Empirical Predictive Model: Tech Employment Forecast (Random Forest Regressor)', fontsize=13)
plt.xlabel('Date', fontsize=11)
plt.ylabel('Tech Employment (Thousands)', fontsize=11)
plt.legend(loc='upper left', frameon=True, facecolor='white')
plt.tight_layout()
plt.show()

# Print Feature Importance
importances = pd.Series(model.feature_importances_, index=feature_cols).sort_values(ascending=False)
print("\n=== TOP PREDICTIVE FEATURE IMPORTANCES ===")
print(importances.head(5).to_string())
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. FIXED STOCHASTIC AI COST DECAY
# ==========================================
np.random.seed(42)
N_trials = 300
T_horizon = 120 # 10 Years
N_agents = 1000

p_T0 = 2.0
mu_T = -0.025   # Realistic 2.5% monthly decay (not instant 99% drop)
sigma_T = 0.04  # 4% monthly volatility

p_T_matrix = np.zeros((N_trials, T_horizon))
p_T_matrix[:, 0] = p_T0

for t in range(1, T_horizon):
    Z = np.random.normal(0, 1, N_trials)
    p_T_matrix[:, t] = p_T_matrix[:, t-1] * np.exp((mu_T - 0.5 * sigma_T**2) + sigma_T * Z)

# Realistic Automation Threshold I*(t) over 10 years
I_star_matrix = 0.90 / (1.0 + (p_T_matrix / p_T0)**0.7)

# ==========================================
# 2. SATURATED MARKOV WORKFORCE TRANSITION
# ==========================================
agent_tasks = np.random.beta(2, 4, N_agents)

# States:
# 0 = Tech Employed
# 1 = Searching / Job Hunt
# 2 = Service Sector (Cap at 30% of total economy)
# 3 = Structural / Permanent Unemployment (Labor Saturation)

SERVICE_CAP = 0.30 * N_agents # Maximum service sector capacity
state_counts = np.zeros((N_trials, T_horizon, 4))

for m in range(N_trials):
    agent_states = np.zeros(N_agents, dtype=int)
    search_time = np.zeros(N_agents)

    for t in range(T_horizon):
        threshold_t = I_star_matrix[m, t]
        current_service_count = np.sum(agent_states == 2)

        for a in range(N_agents):
            s = agent_states[a]

            if s == 0:  # Tech Employed
                if agent_tasks[a] <= threshold_t:
                    agent_states[a] = 1  # Displaced -> Searching
                    search_time[a] = 0

            elif s == 1:  # Searching
                search_time[a] += 1
                # Try entering service sector
                if current_service_count < SERVICE_CAP:
                    if np.random.rand() < 0.15: # 15% monthly hiring probability
                        agent_states[a] = 2
                        current_service_count += 1
                else:
                    # Service sector is FULL (Saturated) -> Worker falls to Structural Unemployment
                    if search_time[a] > 8: # Exhausted benefits after 8 months
                        agent_states[a] = 3

        for s_idx in range(4):
            state_counts[m, t, s_idx] = np.sum(agent_states == s_idx) / N_agents * 100.0

# Calculate Averages Across Trials
mean_tech = np.mean(state_counts[:, :, 0], axis=0)
mean_search = np.mean(state_counts[:, :, 1], axis=0)
mean_service = np.mean(state_counts[:, :, 2], axis=0)
mean_unemployed = np.mean(state_counts[:, :, 3], axis=0)

# ==========================================
# 3. PLOT REALISTIC WORKFORCE DYNAMICS
# ==========================================
months = np.arange(0, T_horizon)
plt.figure(figsize=(12, 6), dpi=120)
plt.style.use('seaborn-v0_8-whitegrid')

plt.stackplot(months, mean_tech, mean_search, mean_service, mean_unemployed,
              labels=[
                  'Tech Sector Employed',
                  'Searching / Frictionally Unemployed',
                  'Service Sector (Saturated at Cap)',
                  'Structural Technological Unemployment (No Jobs Available)'
              ],
              colors=['#2ca02c', '#ff7f0e', '#1f77b4', '#d62728'], alpha=0.85)

plt.title('Realistic Workforce Dynamics: AI Displacement & Labor Saturation Ceiling', fontsize=13)
plt.xlabel('Months Ahead', fontsize=11)
plt.ylabel('Share of Workforce (%)', fontsize=11)
plt.ylim(0, 100)
plt.legend(loc='upper right', frameon=True, facecolor='white')
plt.tight_layout()
plt.show()
```
